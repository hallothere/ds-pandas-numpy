3. I've decided that this month I want to stay away from wines with relatively high alcohol content. To do that, I'm going to avoid any wines that have a greater alcohol content than the mean alcohol content, and you're going to help me do this. To achieve this, let's do the following: 

  * Find the mean alcohol content, separately, for reds and whites.  
  * Create a `Series` that holds whether each row in each `DataFrame` (the reds, and whites) has a higher alcohol content than the mean. 
  * Merge this `Series` onto the `DataFrame`. I can imagine doing this with either a `.join()` or using `pd.concat()`. For practice, do it with both. Note: merges with `Series` work the same way that they work with `DataFrames`.  
  * Return back to me all those rows that will help me stay away from those wines with a higher alcohol content.

```python
#  * Find the mean alcohol content, separately, for reds and whites.  

ave_alc_red = redwine_df["alcohol"].mean()
ave_alc_red

ave_alc_white = whitewine_df["alcohol"].mean()
ave_alc_white
```

```python
#Create a `Series` that holds whether each row in each `DataFrame` (the reds, and whites) has a higher alcohol content than the mean. 

#redwine_df.eval("higher_than_ave = alcohol > alcohol.mean()")
#whitewine_df.eval("higher_than_ave = alcohol > alcohol.mean()")

red_higher_series = redwine_df["alcohol"] > redwine_df["alcohol"].mean()
red_higher_series
white_higher_series = whitewine_df["alcohol"] > redwine_df["alcohol"].mean()
white_higher_series
```

```python
#Merge this `Series` onto the `DataFrame`. I can imagine doing this with either a `.join()` or using `pd.concat()`. 
# For practice, do it with both. Note: merges with `Series` work the same way that they work with `DataFrames`.  

red_higher_series.name = "higher_than_ave"
white_higher_series.name = "higher_than_ave"

red_joined = redwine_df.join(red_higher_series)
red_joined

white_joined = whitewine_df.join(white_higher_series)
white_joined

red_concat = pd.concat([redwine_df, red_higher_series], axis=1)
red_concat
white_concat = pd.concat([whitewine_df, white_higher_series], axis=1)
```

```python
# Return back to me all those rows that will help me stay away from those wines with a higher alcohol content.

red_concat.query("higher_than_ave == False")
white_concat.query("higher_than_ave == False")
```

4. Let's say that I want to get started on cutting back next month. This time, though, I want to focus on staying away from those wines with a high acidity. Specifically, I want to stay away from those wines that are in the highest bin of fixed acidity (highest bin out of 5). You're now going to help me do this. To achieve this, let's do the following: 

 * Separate the rows in each `DataFrame` into 5 equal width bins (not equal to quintiles) based off their fixed acidity. 
 * Merge the resulting `Series` holding these 5 bins onto the original `DataFrame`. I can imagine also doing this with either `.join()` or using `pd.concat()`. Try doing it with both for practice. 
 * Return back to me all those rows that are **not in** the top bin in terms of fixed acidity. 


```python
# Separate the rows in each `DataFrame` into 5 equal width bins (not equal to quintiles) based off their fixed acidity.

#red_concat["fixed_acidity_bins"] = pd.cut(red_concat['fixed acidity'], bins=5, labels=["very low", "low", "medium", "high", "very high"])

#red_concat.head()

#red_joined_bins = red_concat.join(acidity_bins)
#red_joined_bins.head()

acidity_bins_white = pd.cut(white_concat['fixed acidity'], bins=5, labels=["very low", "low", "medium", "high", "very high"])
acidity_bins_white.name = "acidity_bins"

white_joined_bins = white_concat.join(acidity_bins_white)
white_joined_bins

red_concat.head()
```

```python
#Return back to me all those rows that are **not in** the top bin in terms of fixed acidity. 

white_joined_bins.query("acidity_bins != 'very high'")
white_joined_bins.head()
```

 * The average `alcohol` content for those reds above the mean `alcohol` level, by quality.
 * The average `alcohol` content for those whites above the mean `alcohol` level, by quality. 
 
 Use a `pivot table` to solve this. 

```python
red_concat[red_concat["higher_than_ave"]].groupby("quality")["alcohol"].mean()
#white_joined_bins.query("higher_than_ave == True").groupby("quality")["alcohol"].mean()
```

```python
pd.pivot_table(red_concat[red_concat["higher_than_ave"]], values='alcohol', index='quality')
```
